# Edge detection in CV

```python
# read the image
img_bgr = cv2.imread('Couple.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

```python
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap='gray')
```

```python
img_blur = cv2.GaussianBlur(img_gray, (3,3), 0)
plt.imshow(img_blur, cmap='gray')
```

```python
# sobel filter
sobelx = cv2.Sobel(src=img_blur, ddepth=cv2.CV_64F, dx=1, dy=0, ksize=5) # xaxis
sobely = cv2.Sobel(src=img_blur, ddepth=cv2.CV_64F, dx=0, dy=1, ksize=5) # yaxis
sobelxy = cv2.Sobel(src=img_blur, ddepth=cv2.CV_64F, dx=1, dy=1, ksize=5) # both x and y
```

```python
plt.figure(figsize=(18, 5))
plt.subplot(141);plt.imshow(sobelx, cmap='gray');plt.title('Sobel x-axis');
plt.subplot(142);plt.imshow(sobely, cmap='gray');plt.title('Sobel y-axis');
plt.subplot(143);plt.imshow(sobelxy, cmap='gray');plt.title('Sobel xy-axis');
```

```python
# perform global thresholding
retval, img_threshold_sobelx = cv2.threshold(sobelx, 100, 255, cv2.THRESH_BINARY)
retval, img_threshold_sobely = cv2.threshold(sobely, 100, 255, cv2.THRESH_BINARY)
retval, img_threshold_sobelxy = cv2.threshold(sobelxy, 100, 255, cv2.THRESH_BINARY)
```

```python
plt.figure(figsize=[18, 5])
plt.subplot(141);plt.imshow(img_threshold_sobelx, cmap='gray');plt.title('Thresh Sobelx');
plt.subplot(142);plt.imshow(img_threshold_sobely, cmap='gray');plt.title('Thresh Sobely');
plt.subplot(143);plt.imshow(img_threshold_sobelxy, cmap='gray');plt.title('Thresh Sobelxy');
```
