```python
img_path = 'Couple.jpg'
img = cv2.imread(img_path)
```

```python
# load the pre-trained model - Haar cascade classifier for face detection
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
faces = face_cascade.detectMultiScale(gray_img, scaleFactor=1.1, 
                                      minNeighbors=5, minSize=(30,30))
for (x, y, w, h) in faces:
    cv2.rectangle(img, (x, y), (x+w, y+h), (255, 0, 0), 2)

cv2.imshow('Detected Faces', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# sample sentence
text_sample = """He determined to drop his litigation with the monastry, and relinquish his claims to the wood cutting and fishery rights at once. He was more ready to do this because the rights ahad become much less valuable, and he had indeed the vaguest idea where the wood and river in questions were."""
# set the stop words
stop_words = set(stopwords.words('english'))
# tokens
word_tokens = word_tokenize(text_sample)
filtered_sentence = []
for w in word_tokens:
    if w not in stop_words:
        filtered_sentence.append(w)

print('Original Sentence')
print(" ".join(word_tokens))
print('*'*50)
print('Filtered Sentence')
print(" ".join(filtered_sentence))
```

```python
from collections import Counter
from string import punctuation
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
def get_hotwords(text):
    result = []
    pos_tag = ['PROPN', 'ADJ', 'NOUN']
    doc = nlp(text.lower())
    for token in doc:
        if(token.text in nlp.Defaults.stop_words or token.text in punctuation):
            continue
        if(token.pos_ in pos_tag):
            result.append(token.text)
    return result
```

```python
new_text = """When it comes to evaluating the performance of keyword extractors, you can use some of the standard metrics in Machine Learning: accuracy, precision, recall and F1 Score. However, these metrics don't reflect partial matches. They only consider the perfect match between the extracted segments and the correct prediction of the tag.
Fortunately, there are some other metrics capable of capturing partial matches like ROUGE"""
```

```python
output = set(get_hotwords(new_text))
most_common_list = Counter(output).most_common(10)
for item in most_common_list:
    print(item[0])
```

```python
# Stemming
from nltk.stem import PorterStemmer
ps = PorterStemmer()
print(ps.stem("Stemming"))
```

```python
sentence = 'It is good to be a pythoner who is pythonly pythoning with python'
words = nltk.word_tokenize(sentence)
for i in words:
    print(ps.stem(i))
```

```python
# snowball stemmer
from nltk.stem.snowball import SnowballStemmer
sbs = SnowballStemmer(language="english")
word = 'civilization'
sbs.stem(word)
```

```python
for i in words:
    print(sbs.stem(i))
```

```python
# lancaster stemmer - shortest stem possible
from nltk.stem import LancasterStemmer
lanc = LancasterStemmer()
for i in words:
    print(lanc.stem(i))
```

```python
# Lemmatization - normalization - dict 
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
porter = PorterStemmer()
lemmatizer = WordNetLemmatizer()
print(f'better\n Stemming: {porter.stem('better')}\n Lemmatization:{lemmatizer.lemmatize('better', pos='a')}')
```

```python
# sentence
sentence = "There are mistakes"
print(f'Sentence: {sentence}')

word_list = nltk.word_tokenize(sentence)
print(f'word list: {word_list}')

lem_out = ' '.join([lemmatizer.lemmatize(i) for i in word_list])
print(f'Lemma: {lem_out}')
```
