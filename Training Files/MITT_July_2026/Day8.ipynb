```python
# Split the image into B, G, R
img_NZ = cv2.imread('/Users/raghunandanms/Downloads/Spoke Engg/Unit 5-Computer Vision Fundamentals & Applications/Lab/Lab 26 Perform image thresholding, annotation, and image resizing by OpenCV/New_Zealand_Lake.jpg', 
                    cv2.IMREAD_COLOR)
b, g, r = cv2.split(img_NZ)

plt.figure(figsize=(20, 5))
plt.subplot(151);plt.imshow(r, cmap='gray');plt.title('Red Channel');
plt.subplot(152);plt.imshow(g, cmap='gray');plt.title('Green Channel');
plt.subplot(153);plt.imshow(b, cmap='gray');plt.title('Blue Channel');

img_merged = cv2.merge((b, g, r))
plt.subplot(154);plt.imshow(img_merged[:,:,::-1]);plt.title('Merged Output');
```

```python
img_NZ_rgb = cv2.cvtColor(img_NZ, cv2.COLOR_BGR2RGB)
plt.imshow(img_NZ_rgb)
```

```python
cv2.imwrite('RGB_NZ.png', img_NZ_rgb)
```

```python
# Crop the image
img_NZ_rgb_cropped = img_NZ_rgb[200:400, 300:600]
#print(img_NZ_rgb_cropped)
plt.imshow(img_NZ_rgb_cropped)
```

```python
# Resize the image
resized_img = cv2.resize(img_NZ_rgb_cropped, None, fx=2, fy=2)
plt.imshow(resized_img)
```

```python
print('Original Image', img_NZ_rgb_cropped.shape)
print('Resized Image', resized_img.shape)
```

```python
img_NZ_rgb_flipped_horz = cv2.flip(img_NZ_rgb, 1)
img_NZ_rgb_flipped_vert = cv2.flip(img_NZ_rgb, 0)
img_NZ_rgb_flipped_both = cv2.flip(img_NZ_rgb, -1)

# Show the images
plt.figure(figsize=[18,5])
plt.subplot(141);plt.imshow(img_NZ_rgb_flipped_horz);plt.title("Horizontal Flip");
plt.subplot(142);plt.imshow(img_NZ_rgb_flipped_vert);plt.title("Vertical Flip");
plt.subplot(143);plt.imshow(img_NZ_rgb_flipped_both);plt.title("Both Flipped");
plt.subplot(144);plt.imshow(img_NZ_rgb);plt.title("Original");
```
