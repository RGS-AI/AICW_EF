```python
# __init__ constructor - used to initialize
class person:
    def __init__(self, name, age, school):
        self.name = name
        self.age = age
        self.school = school
    def myfun(self):
        print('Hello, My name is', self.name)
        print('My age is', self.age)
        print('I went to', self.school)
```

```python
p1 = person('RMS', 38, 'Geetha School')
p1.myfun()
```

```python
# accessing attributes and methods
class emp:
    name = 'RMS'
    salary = '25000'
    def show(self):
        print(self.name)
        print(self.salary)
```

```python
e1 = emp()
```

```python
# using getattr instead e1.name
print(getattr(e1, 'name'))
```

```python
# hasattr - return true if the object has attribute
print(hasattr(e1, 'name'))
```

```python
print(hasattr(e1, 'gender'))
```

```python
# setattr - sets an attribute
setattr(e1, 'gender', 'M')
```

```python
print(getattr(e1, 'gender'))
```

```python
# delattr - del the attribute
delattr(e1, 'gender')
```

```python
print(getattr(e1, 'gender'))
```

```python
class Student:
    def __init__(self, usn, name, age): # constructor
        # Instance variable
        self.usn = usn
        self.name = name
        self.age = age
    # instance method access instance variable
    def show(self):
        print('USN', self.usn, 'Name', self.name,
             'Age', self.age)
    # instance method to modify instance variable
    def update(self, usn, age):
        self.usn = usn
        self.age = age
```

```python
# create an object
stu = Student(40, 'RMS', 21)
stu.show()
```

```python
stu.update(12, 38)
stu.show()
```
